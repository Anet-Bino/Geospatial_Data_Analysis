```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import folium

print("All libraries imported successfully!")
```

```python

import pandas as pd

data = {
    'City': ['Kochi', 'Kannur', 'Kozhikode', 'Thrissur', 'Trivandrum',
             'Bengaluru', 'Chennai', 'Mumbai', 'Delhi', 'Coimbatore',
             'Kollam', 'Palakkad'],

    'State': ['Kerala', 'Kerala', 'Kerala', 'Kerala', 'Kerala',
              'Karnataka', 'Tamil Nadu', 'Maharashtra', 'Delhi',
              'Tamil Nadu', 'Kerala', 'Kerala'],

    'Latitude': [9.9312, 11.8745, 11.2588, 10.5276, 8.5241,
                 12.9716, 13.0827, 19.0760, 28.6139, 11.0168,
                 8.8932, 10.7867],

    'Longitude': [76.2673, 75.3704, 75.7804, 76.2144, 76.9366,
                  77.5946, 80.2707, 72.8777, 77.2090, 76.9558,
                  76.6141, 76.6548],

    'Sales': [85000, 45000, 70000, 55000, 90000,
              150000, 120000, 180000, 200000, 65000,
              40000, 35000],

    'Customers': [850, 400, 700, 550, 950,
                  1500, 1200, 1800, 2000, 650,
                  350, 300],

    'Store_Count': [8, 3, 5, 4, 9, 15, 12, 18, 20, 5, 2, 2]
}

df = pd.DataFrame(data)

df.to_csv('geospatial_sales.csv', index=False)

print("Dataset created successfully!")
print(df.head())
```

```python

# Display the complete dataset
df
```

```python

# Check dataset shape
print("Dataset Shape:", df.shape)

# Check column names
print("\nColumn Names:")
print(df.columns)

# Check data types and missing values
print("\nDataset Information:")
df.info()
```

```python

# Display the complete dataset
display(df)

# Check number of rows and columns
print("Dataset Shape:", df.shape)

# Check dataset information
df.info()
```

```python

# Check missing values
print("Missing Values:")
print(df.isnull().sum())

# Check duplicate records
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python

# Sort cities based on sales
city_sales = df.sort_values(by='Sales', ascending=False)

print("Sales by City:")
display(city_sales[['City', 'Sales', 'Customers', 'Store_Count']])
```

```python

print("Highest Sales City:")
print(df.loc[df['Sales'].idxmax(), ['City', 'Sales']])

print("\nLowest Sales City:")
print(df.loc[df['Sales'].idxmin(), ['City', 'Sales']])

print("\nTotal Sales:", df['Sales'].sum())
print("Total Customers:", df['Customers'].sum())
```

```python

plt.figure(figsize=(12, 6))

plt.bar(df['City'], df['Sales'])

plt.title('Sales by City')
plt.xlabel('City')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)

plt.show()
```

```python

plt.figure(figsize=(12, 6))

plt.bar(df['City'], df['Customers'])

plt.title('Customer Demand by City')
plt.xlabel('City')
plt.ylabel('Number of Customers')
plt.xticks(rotation=45)

plt.show()
```

```python

# Create a map centered on India
india_map = folium.Map(
    location=[20.5937, 78.9629],
    zoom_start=5
)

# Add city markers
for index, row in df.iterrows():
    folium.Marker(
        location=[row['Latitude'], row['Longitude']],
        popup=f"City: {row['City']}<br>Sales: {row['Sales']}<br>Customers: {row['Customers']}",
        tooltip=row['City']
    ).add_to(india_map)

# Display map
india_map
```

```python

from folium.plugins import HeatMap

sales_map = folium.Map(
    location=[20.5937, 78.9629],
    zoom_start=5
)

heat_data = df[['Latitude', 'Longitude', 'Sales']].values.tolist()

HeatMap(heat_data).add_to(sales_map)

sales_map
```

```python

# Calculate customers per store
df['Customers_Per_Store'] = (
    df['Customers'] / df['Store_Count']
)

# Sort cities by customers per store
expansion = df.sort_values(
    by='Customers_Per_Store',
    ascending=False
)

print("Potential Business Expansion Locations:")

display(
    expansion[['City', 'Customers',
               'Store_Count', 'Customers_Per_Store']]
)
```

```python

plt.figure(figsize=(12, 6))

plt.bar(
    expansion['City'],
    expansion['Customers_Per_Store']
)

plt.title('Business Expansion Opportunities by City')
plt.xlabel('City')
plt.ylabel('Customers per Store')
plt.xticks(rotation=45)

plt.show()
```

```python

# Get top 5 cities for expansion
top_locations = expansion.head(5)

print("Top 5 Potential Expansion Locations:")
display(top_locations[
    ['City', 'State', 'Sales', 'Customers',
     'Store_Count', 'Customers_Per_Store']
])
```

```python

expansion_map = folium.Map(
    location=[20.5937, 78.9629],
    zoom_start=5
)

for index, row in top_locations.iterrows():
    folium.CircleMarker(
        location=[row['Latitude'], row['Longitude']],
        radius=10,
        popup=f"{row['City']} - {row['Customers_Per_Store']:.2f} customers/store",
        tooltip=row['City'],
        color='red',
        fill=True,
        fill_color='red'
    ).add_to(expansion_map)

expansion_map
```

```python

# Save analysis results
df.to_csv('geospatial_analysis_results.csv', index=False)

# Save interactive maps
india_map.save('india_sales_map.html')
sales_map.save('sales_heatmap.html')
expansion_map.save('expansion_opportunities_map.html')

print("All project files saved successfully!")
```
